## 📘 Формулировка задачи

Представьте, что вы работаете аналитиком в HR-отделе компании. Вам нужно смоделировать распределение зарплат сотрудников в трёх отделах (IT, Sales, HR) и построить визуализацию для сравнения.

### Используя numpy, сгенерируйте выборки зарплат:

**IT-отдел: средняя зарплата ≈ 120 000 руб., стандартное отклонение 20 000 руб.**

**Sales-отдел: средняя зарплата ≈ 80 000 руб., стандартное отклонение 15 000 руб.**

**HR-отдел: средняя зарплата ≈ 60 000 руб., стандартное отклонение 10 000 руб.**

**Каждая выборка — 200 сотрудников.**

### Постройте с помощью matplotlib:

**Гистограммы распределения зарплат для трёх отделов на одном графике.**

**Коробчатую диаграмму (boxplot) для сравнения.**

### Вычислите и выведите:

**среднее, медиану и стандартное отклонение для каждого отдела.**

**отдел с наибольшей вариативностью зарплат.**

### 🔹 Python (базовые вещи)

zip() — объединяет несколько списков/массивов в пары для перебора.

Циклы for, форматирование строк f"...".

Работа со списками и массивами.

### 🔹 NumPy (ключевые методы)

np.random.normal(mean, std, size) — генерация случайных данных (нормальное распределение).

np.mean(arr) — среднее значение.

np.median(arr) — медиана.

np.std(arr) — стандартное отклонение (разброс данных).

np.arange(start, stop, step) и np.linspace(start, stop, num) — создание массивов.

Индексация и срезы массивов (например, arr[0:10]).

### 🔹 Matplotlib (минимум)

plt.hist() — гистограмма.

plt.boxplot() — boxplot.

plt.plot() — линейный график.

plt.show() — показать график.

```python
# Подключаем библиотеки:
# numpy — для работы с массивами и генерацией случайных данных,
# matplotlib.pyplot — для построения графиков.
import numpy as np
import matplotlib.pyplot as plt

# Фиксируем "зерно" генератора случайных чисел,
# чтобы каждый запуск кода давал одинаковые результаты (удобно для обучения и отладки).
np.random.seed(42)

# Генерируем зарплаты в IT-отделе:
# распределение нормальное (колокол), среднее = 120 000, стандартное отклонение = 20 000,
# количество сотрудников = 200.
it_salaries = np.random.normal(120_000, 20_000, 200)

# Аналогично генерируем зарплаты в Sales-отделе: среднее 80 000, разброс 15 000.
sales_salaries = np.random.normal(80_000, 15_000, 200)

# Аналогично для HR-отдела: среднее 60 000, разброс 10 000.
hr_salaries = np.random.normal(60_000, 10_000, 200)

# ---------- ВИЗУАЛИЗАЦИЯ №1: ГИСТОГРАММЫ ----------

# Создаём окно графика с размером 10 на 6 дюймов.
plt.figure(figsize=(10, 6))

# Строим гистограмму зарплат в IT, разбиваем диапазон на 20 интервалов (bins),
# делаем прозрачность 0.6, добавляем подпись "IT".
plt.hist(it_salaries, bins=20, alpha=0.6, label="IT")

# То же для Sales.
plt.hist(sales_salaries, bins=20, alpha=0.6, label="Sales")

# То же для HR.
plt.hist(hr_salaries, bins=20, alpha=0.6, label="HR")

# Добавляем заголовок графика.
plt.title("Распределение зарплат сотрудников")

# Подписываем ось X — это зарплата в рублях.
plt.xlabel("Зарплата, руб.")

# Подписываем ось Y — количество сотрудников в каждом интервале зарплат.
plt.ylabel("Количество сотрудников")

# Добавляем легенду (чтобы различать отделы по цветам).
plt.legend()

# Показываем построенный график.
plt.show()

# ---------- ВИЗУАЛИЗАЦИЯ №2: BOXPLOT ----------

# Создаём окно графика размером 8 на 5 дюймов.
plt.figure(figsize=(8, 5))

# Строим boxplot (коробчатую диаграмму) для трёх отделов.
# Это позволяет сравнить медиану, квартили и выбросы.
plt.boxplot([it_salaries, sales_salaries, hr_salaries], labels=["IT", "Sales", "HR"])

# Добавляем заголовок.
plt.title("Сравнение зарплат по отделам (boxplot)")

# Подписываем ось Y.
plt.ylabel("Зарплата, руб.")

# Показываем график.
plt.show()

# ---------- РАСЧЁТ СТАТИСТИК ----------

# Перебираем пары "название отдела" и "массив данных"
for dept, data in zip(["IT", "Sales", "HR"], [it_salaries, sales_salaries, hr_salaries]):
    # Выводим среднее, медиану и стандартное отклонение (округляем до целого)
    print(f"{dept}: Среднее={np.mean(data):.0f}, "
          f"Медиана={np.median(data):.0f}, "
          f"Std={np.std(data):.0f}")
```

## 📝 Задача для самостоятельной работы

Ситуация:
Вы исследуете цифровое поведение студентов. Нужно смоделировать данные о том, сколько времени в среднем студенты проводят в социальных сетях в день, и сравнить факультеты.

### С помощью numpy сгенерируйте выборки для трёх факультетов (по 250 студентов в каждом):

Факультет информатики: среднее = 3.5 часа, std = 1.5.

Экономический факультет: среднее = 2.5 часа, std = 1.0.

Филологический факультет: среднее = 2.0 часа, std = 0.8.

### С помощью matplotlib постройте:

Гистограммы для трёх факультетов на одном графике.

Boxplot для сравнения.

### Рассчитайте:

среднее, медиану и стандартное отклонение для каждого факультета.

процент студентов, проводящих в соцсетях более 4 часов в день.

```python
# Задача 1

mosc = np.random.normal(120000, 25000, 300)
sanp = np.random.normal(90000, 20000, 300)
kaz = np.random.normal(70000, 15000, 300)

towns=["Moscow", "Peterburg", "Kazan"]
#Hist

plt.figure(figsize=(10, 6))

plt.hist(mosc, bins=20, alpha=0.6, label="Moscow")
plt.hist(sanp, bins=20, alpha=0.6, label="Peterburg")
plt.hist(kaz, bins=20, alpha=0.6, label="Kazan")

plt.title("Распределение расходов в городах")
plt.xlabel("Расходы, руб.")
plt.ylabel("Семей")
plt.legend()

plt.show()

# BoxPlot

plt.figure(figsize=(8, 5))

plt.boxplot([mosc, sanp, kaz], labels=towns)


plt.title("Сравнение расходов по городам (boxplot)")
plt.ylabel("Расходы, руб.")

plt.show()

# Рассчёты

for town, arr in zip(towns, [mosc, sanp, kaz]):
    print('------' + town + '------')
    print(f'Среднее: {round(np.mean(arr))}', f'Медиана: {round(np.median(arr))}', f'Ст. откл: {round(np.std(arr))}', sep='\n')
    count = 0
    for i in arr:
        if i>100000:
            count+=1
    print(f'Процент семей с расходами > 100000 руб: {round((count/300)*100)} %\n\n')
```

```python
# Задача 2

inf = np.random.normal(3.5, 1.5, 250)
eko = np.random.normal(2.5, 1, 250)
fil = np.random.normal(2, 0.8, 250)

facults=["Inform", "Ekonom", "Filol"]
#Hist

plt.figure(figsize=(10, 6))

plt.hist(inf, bins=20, alpha=0.6, label="Inform")
plt.hist(eko, bins=20, alpha=0.6, label="Ekonom")
plt.hist(fil, bins=20, alpha=0.6, label="Filol")

plt.title("Распределение времени проведения в соц сетях по факультетам")
plt.xlabel("Время, ч.")
plt.ylabel("Человек")
plt.legend()

plt.show()

# BoxPlot

plt.figure(figsize=(8, 5))

plt.boxplot([inf, eko, fil], labels=towns)


plt.title("Распределение времени проведения в соц сетях по факультетам (boxplot)")
plt.ylabel("Время, ч.")

plt.show()

# Рассчёты

for facult, arr in zip(facults, [inf, eko, fil]):
    print('------' + facult + '------')
    print(f'Среднее: {np.mean(arr)}', f'Медиана: {np.median(arr)}', f'Ст. откл: {np.std(arr)}', sep='\n')
    count = 0
    for i in arr:
        if i>4:
            count+=1
    print(f'Процент человек, который сидит в соц сетях больше 4 часов: {round((count/250)*100)} %\n\n')
```
